```r
for (i in c(2, 4, 6)) { print(i^2)}
```

```r
x = c(2, 4, 8, 10)

excount = function(x) {
  count = 0
  for (xval in x)
  {
    if(xval/2 > 3)
    {
      count = count+1
    }
  }
  print(count)
}
excount(x)
```

```r
# nested looping for loop

child = c("child1", "child2")
sweet = c("sweet1", "sweet2")

for(x in child)
{
  for (y in sweet)
  {
    print(paste(x,y))
  }
}
```

```r
# break command

drink = c("coffee", "tea", "juice")

for(x in drink)
{
  if(x == "tea") break
  print(x)
}
```

```r
# next command

drink = c("coffee", "tea", "juice")

for(x in drink)
{
  if(x == "tea") next
  print(x)
}
```

```r
# while loop

i = 1
while(i<10)
{
  print(i^2)
  i = i+2
}
```

```r
sumfunc = function(){
  sum = 0
  num = as.integer(readline(prompt="Please select any number less than 25: "))
  while (num<=25)
  {
    sum = sum + num
    num = num + 1
  }
  print(paste("Sum of all numbers is ", sum))
}
sumfunc()
```

```r
# repeat loop

i = 1
repeat{
  print(i^2)
  i = i+2
  if( i>10)
  break
}
```

```r
# function

squarenum = function(x)
{
  x^2
}
squarenum(5)
```

```r
# function with 2 variables

abc = function(x, y)
{
  sin(x) + cos(y)^2 + x
}
abc(2, 5)
```

```r
# sequences

#default increment is +1 or -1
seq(from=2, to=5)
seq(from=2, to=-2)
```

```r
seq(from=1, to=5, by=2)
seq(from=1, length=10, by=0.2)
```
