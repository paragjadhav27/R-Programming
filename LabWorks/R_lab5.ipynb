```r
#Merging lists
list1 = list(3,2,1,4)
list2 = list("apple", "juice", "lemon", 2.5)
list12 = c(list1, list2)
list12
```

```r
#Lists to vector
unlist(list1)
unlist(list2)

mode(list1)
mode(list2)
mode(unlist(list1))
mode(unlist(list2))
mode(unlist(list12))
```

```r
#Appending lists
append(list1, 100)
append(list2, "coffee")

append(list1, 100, after=3)
```

```r
#Removing from list
list1[-1]
list2[-4]
```

```r
#Extracting from list
list1[2:4]
list1[c(1,4)]
```

```r
#Vector Indexing
x = 1:10
x

x[x%%2 == 1]
x[x>4 & x<8]

x[6:10]
x[-(1:5)] #vector of negative integers
```

```r
#Vector with missing observations
x[5] = NA
x

y = x[!is.na(x)]
y

mean(x)
mean(y)
```

```r
#Heterogeneous (Mixed mode)
ab = list(1, 2, 3, "X", "Y", "Z")
dim(ab) = c(2, 3)
mode(print(ab))
```

```r
#Factors
y = c(5, 10, 15, 20)
y

lev = c(5, 10, 15, 20)
lab = c("less than 10", "10", "greater than 10", "greater than 10")

facy = factor(y, levels = lev, labels = lab)
facy
facty = as.factor(y)
facty
```

```r
x = c(3, 4, 5, 6, 8, 1, 1, 3)
y = factor(x)
```

```r
#Factors
#class function

class(9)
class("9")
class(print)
x = matrix(nrow=2, ncol=2, data=1:4)
class(x)
x = matrix(c(1, "AB", 2, "B"), nrow = 2, ncol = 2)
class(x)
```

```r
#unclass function

colors = c("green", "red", "yellow", "red", "yellow")
colors_fac = factor(colors)
colors_fac

unclass(colors_fac)
attr(colors_fac ,"levels")
```

```r
#Ordered factor

income = ordered(c("high", "high", "low", "medium", "high"), levels=c("low", "medium", "high"))
income

unclass(income)
attr(income ,"levels")
```
